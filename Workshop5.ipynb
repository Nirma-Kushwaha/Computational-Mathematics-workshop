Example 7:
Create two vectors in 2-dimensional (2D) space and plot them using Matplotlib.

Solution: A vector in 2D space can created as a 1D ndarray with two elements as:

V = np.array([1,1])
W = np.array([-2.0,-1])

To plot the above vectors, we will use the quiver() function from the pyplot module of Matplotlib
library. The quiver() function is used to create 2D field or velocity plots and is perfect for our purpose
of vector plotting. Here is an example of plotting the above vectors.

```python
import numpy as np
import matplotlib.pyplot as plt
""" Problem 2: Create 2-dimensional position vectors and plot them. """
origin = np.array([0,0])
V = np.array([1,1])
W = np.array([-2.0,-1])
fig, ax = plt.subplots()
# Add vectors V and W to the plot
ax.quiver(origin[0], origin[1], V[0], V[1], angles='xy', scale_units='xy', scale=1, color='r')
ax.quiver(origin[0], origin[1], W[0], W[1], angles='xy', scale_units='xy', scale=1, color='b')
ax.text(origin[0]-0.1, origin[1]+0.1, "O")
ax.text(V[0], V[1], "v")
ax.text(W[0]-0.125, W[1]-0.1, "w")
ax.set_xlim([-3, 3])
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.grid()
plt.show()
```

a. Create three vectors in 3D space and plot them using Matplotlib.

U= np.array([1, 0.25 ,0])

V = np.array([0, 1, 0.25])

W = np.array([0.25, 0, 1])

To do 3D plot, you will need to create 3D axes using following code:

fig = plt.figure()

ax = fig.add_subplot(111, projection='3d')

```python
import numpy as np
import matplotlib.pyplot as plt
#vectors
U= np.array([1, 0.25 ,0])
V = np.array([0, 1, 0.25])
W = np.array([0.25, 0, 1])
#create 3D axes
fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')
#origin
origin = [0, 0, 0]
#plot vectors
ax.quiver(*origin, *U, color='r', label='U')
ax.quiver(*origin, *V, color='g', label='V')
#labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.legend()
plt.show()
```

b. Create two vectors in 2D space, and add them to create a third vector, and plot all three vectors
using Matplotlib.

U = np.array([-0.5, 0.5])

V = np.array([1,1])

The two vectors are added as

W = U + V # adding two vector

```python
import numpy as np
import matplotlib.pyplot as plt
#vectors
U = np.array([-0.5, 0.5])
V = np.array([1, 1])
#addition
W = U + V
#origin
origin = [0, 0]
#plot vectors
plt.quiver(*origin, *U, angles='xy', scale_units='xy', scale=1, color='r', label='U')
plt.quiver(*origin, *V, angles='xy', scale_units='xy', scale=1, color='g', label='V')
plt.quiver(*origin, *W, angles='xy', scale_units='xy', scale=1, color='b', label='W = U +V')
#axis setting
plt.xlim(-1, 2)
plt.ylim(-1, 2)
plt.axhline(0)
plt.axvline(0)
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid()
plt.show()
```

c. Write a function to calculate magnitude of a vector. For example, calculate the magnitude of
the vector be U = (1.5, 1.5, 2.0).

```python
import numpy as np
def magnitude(v):
  return np.sqrt(np.sum(v**2))
#vector
U = np.array([1.5, 1.5, 2.0])
#magnitude
mag = magnitude(U)
print("Magnitude of U =", mag)
```

d. Write a function to calculate the direction cosines of the vector. For example, let the vector be
O = (1.5, 1.5, 2.0)

```python
import numpy as np
def direction_cosines(v):
  mag = np.sqrt(np.sum(v**2))
  return v/mag
#vector
o = np.array([1.5, 1.5, 2.0])
#direction cosines
dc = direction_cosines(o)
print("Direction cosines=", dc)
```

e. Write a function to calculate Scalar (dot) product of two vectors. For example, let the vectors
be a= (1.5, 1.5, 2.0) and b = (−1.5, 0.5, 2.0).

```python
import numpy as np
def dot_product(a, b):
  return np.sum(a*b)
#vectors
a = np.array([1.5, 1.5, 2.0])
b = np.array([-1.5, 0.5, 2.0])
#dot product
dp = dot_product(a, b)
print("Dot product =", dp)
```

f. Write a function to calculate Vector (cross) product of two vectors. For example, let the vectors
be a = (1.5, 1.5, 2.0) and b= (−1.5, 0.5, 2.0).

```python
import numpy as np
def cross_product(a, b):
  return np.cross(a, b)
#vectors
a = np.array([1.5, 1.5, 2.0])
b = np.array([-1.5, 0.5, 2.0])
#cross product
cp = cross_product(a,b)
print("Cross product =", cp)
```
